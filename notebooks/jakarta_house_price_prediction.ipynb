## 2.4 Korelasi Antar Fitur & Deteksi Outlier

```python
# Korelasi fitur numerik terhadap target
numeric_columns = data_explore.select_dtypes(include="number").columns
correlation = data_explore[numeric_columns].corr(numeric_only=True)
plt.figure(figsize=(11, 8))
sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Matriks Korelasi Variabel Numerik")
plt.tight_layout()
plt.show()

display(correlation["price"].drop("price").sort_values(key=abs, ascending=False).to_frame("Korelasi dengan Target"))
```

```python
# Pemeriksaan outlier berbasis IQR
outlier_summary = []
for column in numeric_columns:
    q1, q3 = data_explore[column].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    count = ((data_explore[column] < lower) | (data_explore[column] > upper)).sum()
    outlier_summary.append({"column": column, "IQR": round(iqr, 2), "Batas Bawah": round(lower, 2), "Batas Atas": round(upper, 2),"Jumlah Outlier": count, "Persentase Outlier": round(count / len(data_explore) * 100, 2)})

display(pd.DataFrame(outlier_summary).sort_values("Jumlah Outlier", ascending=False))

print("Insight awal:")
print(f"- Korelasi terkuat dengan harga: {correlation['price'].drop('price').abs().idxmax()}.")
print(f"- Kota dengan median harga tertinggi: {data_explore.groupby('city')['price'].median().idxmax()}.")
print(f"- Kolom dengan missing value: {quality[quality['missing'] > 0].index.tolist()}.")
```

```python
# Hapus Anomali properti (Tanpa Bedrooms dan Bathrooms)
clean_data = clean_data[(clean_data["bed_rooms"] > 0) & (clean_data["bath_rooms"] > 0)]
print(f"Setelah Hapus Anomali (Filtering): {len(clean_data):,} baris")
```

```python
clean_data = clean_data.reset_index(drop=True)
print(f"Total Data Bersih: {len(clean_data):,} baris")

display(clean_data.head())
```

# Data Preprocessing

```python
encoded_data = clean_data.copy()

encoded_data.head()
```

```python
encoded_data = encoded_data[
    (encoded_data["bed_rooms"] >= 1) & (encoded_data["bed_rooms"] <= 15) &
    (encoded_data["bath_rooms"] >= 1) & (encoded_data["bath_rooms"] <= 15) &
    (encoded_data["land_area"] >= 15) &
    (encoded_data["building_area"] >= 15)
]

train_data, test_data = train_test_split(encoded_data, test_size=0.2, random_state=42)
print("Train:", train_data.shape, "| Test:", test_data.shape)

for col in ["price", "land_area", "building_area"]:
    batas_atas = train_data[col].quantile(0.99)
    train_data = train_data[train_data[col] <= batas_atas]
    test_data = test_data[test_data[col] <= batas_atas]

train_harga_per_m2 = train_data["price"] / train_data["land_area"]
test_harga_per_m2 = test_data["price"] / test_data["land_area"]
low_p, high_p = train_harga_per_m2.quantile([0.005, 0.995])

train_data = train_data[(train_harga_per_m2 >= low_p) & (train_harga_per_m2 <= high_p)].copy()
test_data = test_data[(test_harga_per_m2 >= low_p) & (test_harga_per_m2 <= high_p)].copy()

print("\nJumlah data setelah preprocessing:")
print("Train:", train_data.shape, "| Test:", test_data.shape)
```

```python
# Encoding kolom teks 'city' dan 'district'
city_train = pd.get_dummies(train_data["city"], prefix="city", drop_first=True, dtype=int)
city_test = pd.get_dummies(test_data["city"], prefix="city", drop_first=True, dtype=int)

train_data = pd.concat([train_data.drop(columns=["city"]), city_train], axis=1)
test_data = pd.concat([test_data.drop(columns=["city"]), city_test], axis=1)

te = TargetEncoder(target_type="continuous", random_state=42)

y_train_log = np.log1p(train_data["price"].to_numpy().ravel())

train_data["district_encoded"] = te.fit_transform(train_data[["district"]], y_train_log)
test_data["district_encoded"] = te.transform(test_data[["district"]])

train_data = train_data.drop(columns=["district"])
test_data = test_data.drop(columns=["district"])

train_data.columns = [col.lower().replace('<', '_').replace(' ', '_') for col in train_data.columns]
test_data.columns = [col.lower().replace('<', '_').replace(' ', '_') for col in test_data.columns]
```

```python
print("TRAIN DATA")
display(train_data.head())

print("\nTEST DATA")
display(test_data.head())
```

# Feature Engineering

```python
skewed_cols = ["land_area", "building_area"]

def buat_fitur(df):
    fe_data = df.copy()

    # Buat fitur baru
    fe_data["building_to_land_ratio"] = fe_data["building_area"] / fe_data["land_area"]
    fe_data["total_rooms"] = fe_data["bed_rooms"] + fe_data["bath_rooms"]
    fe_data["area_per_room"] = fe_data["building_area"] / fe_data["total_rooms"]

    # Log transformation untuk kolom yang miring/skewed
    for col in skewed_cols:
        fe_data[f"log_{col}"] = np.log1p(fe_data[col])

    fe_data = fe_data.replace([np.inf, -np.inf], np.nan).dropna()
    return fe_data

train_fe = buat_fitur(train_data)
test_fe = buat_fitur(test_data)

X_train = train_fe.drop(columns=["price"] + skewed_cols)
y_train = train_fe["price"]

X_test = test_fe.drop(columns=["price"] + skewed_cols)
y_test = test_fe["price"]

print("Fitur yang dipakai:")
print(X_train.columns.tolist())
print("Train:", X_train.shape, "| Test:", X_test.shape)
```

# Experimentation

```python
models = {
    "LR": LinearRegression(),
    "RFR": RandomForestRegressor(n_jobs=-1),
    "XGBR": XGBRegressor()
}
```

```python
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    print(f"{name} | R2: {r2:.4f} | MAE: ${mae:,.0f} | RMSE: ${rmse:,.0f}")
```

# Hyperparameter Tuning RandomForest & XGBoost (Perbandingan)

```python
param_dist = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 20, None],
    'min_samples_leaf': [1, 2, 4],
    'max_features': [None, 'sqrt', 0.7]
}

RFR = RandomForestRegressor(random_state=42)

random_search = RandomizedSearchCV(
    estimator=RFR,
    param_distributions=param_dist,
    n_iter=10,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train, y_train)

print("Best Params:", random_search.best_params_)

best_rfr = random_search.best_estimator_
y_pred = best_rfr.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE  : Rp{mae:,.0f}")
print(f"RMSE : Rp{rmse:,.0f}")
print(f"R²   : {r2:.4f}")
```

```python
param_dist_xgb = {
    'n_estimators': [200, 300, 500],
    'learning_rate': [0.03, 0.05, 0.1],
    'max_depth': [4, 6, 8],
    'subsample': [0.7, 0.8, 1.0],
    'colsample_bytree': [0.7, 0.8, 1.0],
    'min_child_weight': [1, 3, 5]
}

xgb = XGBRegressor(random_state=42, n_jobs=-1)

random_search_xgb = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_dist_xgb,
    n_iter=15,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)

random_search_xgb.fit(X_train, np.log1p(y_train))

print("Best Params:", random_search_xgb.best_params_)

best_xgb = random_search_xgb.best_estimator_

# Kembalikan prediksi ke skala Rupiah asli
y_pred = np.expm1(best_xgb.predict(X_test))

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE  : Rp{mae:,.0f}")
print(f"RMSE : Rp{rmse:,.0f}")
print(f"R²   : {r2:.4f}")
```

# Final Model

```python
xgb_model = XGBRegressor(
    n_estimators=500,
    learning_rate=0.03,
    max_depth=6,
    min_child_weight=5,
    subsample=0.7,
    colsample_bytree=0.7,
    reg_lambda=3,
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train, np.log1p(y_train))

y_pred = np.expm1(xgb_model.predict(X_test))
y_train_pred = np.expm1(xgb_model.predict(X_train))

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE      : Rp{mae:,.0f}")
print(f"RMSE     : Rp{rmse:,.0f}")
print(f"Train R² : {r2_score(y_train, y_train_pred):.4f}")
print(f"Test R²  : {r2:.4f}")
```
